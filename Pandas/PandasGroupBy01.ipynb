```python
import pandas as pd

# Load the Excel file into a DataFrame
# Note: Ensure the file path is correct for your environment (e.g., Google Colab or local setup)
# This assumes 'EMP.xlsx' contains employee data with columns like 'SALARY' and possibly others (e.g., employee names, departments)
# If the file has multiple sheets, specify sheet_name if needed; otherwise, it loads the first sheet by default
df = pd.read_excel('/content/EMP.xlsx')



# Group the DataFrame by the 'SALARY' column
# This creates a GroupBy object where rows are grouped by unique salary values
# Note: GroupBy objects are iterable and can be used for aggregation, transformation, or iteration over groups
df1 = df.groupby('SALARY')
print(df1)
#print(df1)  # Uncomment to print the GroupBy object (shows grouping info but not the data)

# Iterate over each group in the GroupBy object
# For each unique 'SALARY' value, 'rows' will be a DataFrame containing all rows with that salary
# This is useful for inspecting or processing each group separately
for SALARY, rows in df1:
    print(f"Salary: {SALARY}")  # Print the salary value for the current group
    print(rows)                 # Print the DataFrame subset for this group
    print("\n")                 # Add a newline for separation between groups

# Perform a groupby on 'SALARY' and apply the count() aggregation
# This counts the number of non-null values in each column for each unique 'SALARY'
# The result is a DataFrame with 'SALARY' as the index and counts for other columns
# Note: If you want 'SALARY' as a regular column, use groupby(as_index=False)
# This is useful for getting frequency counts of salaries or seeing how many entries per salary
df2 = df.groupby('SALARY').count()
print(df2)
```
