```python
from os import cpu_count
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt


pd.set_option('display.max_columns',None)
pd.set_option('display.width',1500)
sns.set_theme(style='darkgrid')
mpg=sns.load_dataset('mpg')

mpg['origin'].unique()


print('\n Output of Head \n',mpg.head())
print('\n Output of Shape \n',mpg.shape)
print('\n Origin of the ountry  \n',mpg['origin'].unique() ,'\n\n')

sns.countplot(x='cylinders' , data=mpg , palette='dark:brown' , hue='origin')

plt.legend(loc='upper center')
plt.title("Count of Cylinders")
plt.xlabel("No of Cylinders")
plt.ylabel("Mileage")
plt.figure(figsize=(15, 12))
plt.show()

```
