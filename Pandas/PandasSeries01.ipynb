```python
import pandas as pd

# Load the Excel file into a DataFrame named 'df'
df = pd.read_excel('/content/EMP.xlsx')

# Display the first 5 rows of the original dataset to inspect the structure
print(df.head(5))

# Extract the 'EMPNAME' column as a pandas Series (a 1D array)
print('Working with series ')
ser = pd.Series(df['EMPNAME'])
print(ser)

# Extract the 'EMPID' column as a pandas Series
ser1 = pd.Series(df['EMPID'])
print(ser1)

# Extract the 'SALARY' column as a pandas Series
ser2 = pd.Series(df['SALARY'])
print(ser2)

# Combine (concatenate) the three Series horizontally (side-by-side)
# axis=1 means join columns; axis=0 would stack them vertically
df1 = pd.concat([ser, ser1, ser2], axis=1)
print(df1)

# Rename the columns of the new DataFrame to be more descriptive
df1.columns = ['Employee Name', 'Employee ID', 'Employee SALARY']
print(df1)
```
