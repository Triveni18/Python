```python
# Import the numpy library and alias it as 'np' for convenience
import numpy as np

# Create a numpy array 'arr1' containing the integers 100, 110, 120, 130, 140, and 150, with dtype specified as int
arr1 = np.array([100, 110, 120, 130, 140 , 150], int)

# Print a newline, followed by the string "Content of arr1", another newline, and then the contents of the array
print('\n Content of arr1 \n',arr1)

# Create a numpy array 'arr2' containing the floating-point numbers 100, 110.50, 120.75, 130.25, 140.00, and 150.50, with dtype specified as float
arr2 = np.array([100, 110.50 , 120.75 , 130.25 , 140.00 , 150.50], float)

# Print a newline, followed by the string "Content of arr2", another newline, and then the contents of the array
print('\n Content of arr2 \n',arr2)

# Create a numpy array 'arr3' containing the strings 'Delhi', 'Lucknow', 'Kanpur', 'Kolkatta', and 'Mumbai', with dtype specified as str (note: 'Kolkatta' appears to be a typo and is likely intended as 'Kolkata')
arr3 = np.array(['Delhi' ,'Lucknow' ,'Kanpur','Kolkatta' , 'Mumbai'] , dtype=str)

# Print a newline, followed by the string "Content of arr3", another newline, and then the contents of the array
print('\n Content of arr3 \n',arr3)
```
