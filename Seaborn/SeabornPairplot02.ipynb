```python
# Import necessary libraries:
# - pandas for data manipulation (though not directly used here beyond potential future needs)
# - matplotlib.pyplot for displaying plots
# - seaborn for data visualization and loading datasets
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the Iris dataset from seaborn's built-in datasets
# This dataset contains measurements of iris flowers from three species: setosa, versicolor, and virginica
iris = sns.load_dataset('iris')

# Print the first few rows of the dataset to inspect its content
print('\n Content of iris data set \n ', iris.head())

# Print a header for the first pair plot
print('\n Pair Plot based on iris Data Set \n ')

# Create a pair plot using seaborn's pairplot function
# - iris: The dataset to use
# - hue='species': Colors the points by species for differentiation
# This plot shows pairwise relationships between variables in the dataset, with scatterplots on off-diagonals and histograms/kde on diagonals
sns.pairplot(iris, hue='species')

# Display the plot using matplotlib
plt.show()

# Print a header for the second pair plot with custom palette and markers
print('\n Pair Plot based on iris Data Set with  palette and markers \n ')

# Create another pair plot with additional customizations
# - hue='species': Colors the points by species
# - palette='husl': Uses the 'husl' color palette for better color distinction
# - markers=['o','d','s']: Specifies different marker styles for each species ('o' for circle, 'd' for diamond, 's' for square)
sns.pairplot(iris, hue='species', palette='husl', markers=['o','d','s'])

# Display the plot
plt.show()

# Print a header for the third pair plot with selected variables
print('\n Pair Plot based on iris Data Set with vars \n ')

# Create a pair plot limited to specific variables
# - vars=['sepal_length','sepal_width']: Only includes these two variables in the pairwise comparisons
# - hue='species': Colors the points by species
# This results in a smaller grid focusing on relationships between sepal_length and sepal_width
sns.pairplot(iris, vars=['sepal_length','sepal_width'], hue='species')

# Display the plot
plt.show()
```
