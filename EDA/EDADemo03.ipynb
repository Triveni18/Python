```python
# =============================================================================
# PURPOSE: Exploratory Data Analysis (EDA) setup for the WineQT dataset
# This code:
#   1. Imports pandas
#   2. Configures pandas display settings for large DataFrames (so nothing gets truncated)
#   3. Loads the Wine Quality dataset
#   4. Quickly inspects shape, data types, missing values, statistics, and column names
# =============================================================================

import pandas as pd                     # pandas is the go-to library for data manipulation and analysis

# ------------------- Configure pandas display options -------------------
# These settings prevent pandas from truncating output when printing large tables
pd.set_option('display.width', 1500)      # wider console output
pd.set_option('display.max_columns', 1500) # show all columns
pd.set_option('display.max_rows', 1500)    # show all rows (useful for small/medium datasets)

# ------------------- Load the dataset -------------------
# Reading the CSV file into a pandas DataFrame
# Path is typical for Google Colab (/content/ folder)
df = pd.read_csv('/content/WineQT.csv')

# ------------------- Basic inspection -------------------
print(df.shape)                           # Returns a tuple (rows, columns) → quick size check

print(df.info())                          # Shows:
                                          # - total rows & columns
                                          # - column names & data types
                                          # - non-null count (helps spot missing values)

print(df.describe().T)                    # Statistical summary (count, mean, std, min, max, quartiles)
                                          # .T transposes the output so each column becomes a row
                                          # (much easier to read when there are many features)

print('\n ')                              # Just adds a blank line for cleaner separation in notebook output

# ------------------- Column names -------------------
print(df.columns.to_list)                 # Note: This should ideally be df.columns.tolist() or df.columns.to_list()
                                          # Without () it prints the method object itself.
                                          # Fix suggestion: print(df.columns.tolist())
```
