```python
# Import the NumPy library and give it the alias 'np' for easier typing
import numpy as np

# Create an array of 10 evenly spaced numbers starting at 2 and ending at 20
# linspace stands for "linear space"
arr = np.linspace(2, 20, 10)

# Get the total number of elements (length) of the array
n = len(arr)

# Print the length of the array with some formatting for readability
print('\n Length of array arr \n ', n)

# Print a header for the array contents
print("\n Content of array arr \n")

# Loop through the array using index-based iteration (0 to n-1)
for i in range(n):
    # Print the element located at the current index i
    print(arr[i])
```
