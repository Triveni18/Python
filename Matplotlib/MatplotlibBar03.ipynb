```python
import pandas as pd  # Import pandas for data manipulation
import matplotlib.pyplot as plt  # Import matplotlib's pyplot for creating visualizations

# Load the Excel file into a pandas DataFrame
# The file path is '/content/EmpDetails.xlsx' and we're reading from 'Sheet1'
df = pd.read_excel('/content/EmpDetails.xlsx', sheet_name='Sheet1')

# Print the DataFrame to display the employee data
print("\n Data of Employee Table \n ", df)

# Extract employee IDs and salaries for the x and y axes
x = df['empid']  # x-axis: Employee IDs
y = df['salary']  # y-axis: Employee Salaries

# Create a vertical bar chart
# 'bars' captures the bar container for later labeling
# Customize bar color, edge color, width, and linewidth
bars = plt.bar(x, y, label='Employee Data', color='red', edgecolor='darkgreen',
               width=0.5, linewidth=3.5)

# Add labels on top of each bar with padding to avoid overlapping
plt.bar_label(bars, padding=5)

# Set the plot title and axis labels
plt.title("Microsoft INC")
plt.xlabel("Employees Id")
plt.ylabel("Employees Salary")

# Adjust the y-axis limit to add extra space at the top for labels
plt.ylim(0, max(y) * 1.1)

# Display the vertical bar chart
plt.show()

# Prepare data for horizontal bar chart
# Convert employee IDs to strings to treat them as categorical labels
y_axis = df['empid'].astype(str)
x_axis = df['salary']  # x-axis: Salaries (now horizontal)

# Create a horizontal bar chart using barh()
# Note: 'height' corresponds to 'width' in vertical bars
# Add hatch pattern for visual distinction
bars = plt.barh(y_axis, x_axis, label='Employee Data', color='red',
                edgecolor='darkgreen', height=0.5, linewidth=3.5, hatch='.')

# Add labels to the right of each bar with padding
# Format labels as currency with commas and no decimals
plt.bar_label(bars, padding=10, fmt='${:,.0f}')

# Set the plot title and swap axis labels for horizontal orientation
plt.title("Microsoft INC")
plt.xlabel("Employees Salary")
plt.ylabel("Employees Id")

# Adjust the x-axis limit to add extra space for labels
plt.xlim(0, max(x_axis) * 1.2)

# Display the horizontal bar chart
plt.show()

```
