```python
import pandas as pd

# 1. Load the data
df = pd.read_excel('/content/EMP.xlsx', sheet_name='Sheet1')

print("Original Index:", df.index)
print(df.head())

# 2. Accessing a row by integer index
print("\nRow at index 3:")
print(df.loc[3])

# 3. Setting index (returning a new object)
# This works perfectly; df1 is the new version, df remains unchanged.
df1 = df.set_index('EMPID')
print("\nNew DataFrame (df1) with EMPID index:")
print(df1.head())

# 4. Setting index (In-place)
# FIX: Do not assign df.set_index(inplace=True) to a variable.
df.set_index('EMPID', inplace=True)


# Now 'df' itself is modified.
print("\nOriginal DataFrame (df) after inplace=True:")
print(df.head())

print("\nOriginal DataFrame (df) after reseting index inplace=True:")
df.reset_index(inplace=True)
print(df.head())
```
