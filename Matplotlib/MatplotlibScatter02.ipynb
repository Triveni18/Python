```python
import matplotlib.pyplot as plt

age =[6,8,9,2,10,2,9,4,11,12,9]
price=[97, 87, 82, 120, 77, 122, 99, 77, 74 , 79 , 83]

plt.scatter(age , price , color='red' , marker='P')

plt.title('Age vs Price')
plt.xlabel('Age')
plt.ylabel('Price')

plt.grid(True)
plt.show()
```
