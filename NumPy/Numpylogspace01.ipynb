```python
# Import necessary libraries: numpy for numerical operations and pandas for data manipulation (though pandas is not used here)
import numpy as np
import pandas as pd

# Create an array 'a' with 4 logarithmically spaced values from 10^1 to 10^4 (default base=10)
a = np.logspace(1, 4, 4)

# Print the content of array 'a'
print('\n Content of Array a  \n ', a)

# Get the length of array 'a' (which is 4)
n = len(a)

# Loop through each element in 'a' and print it with 2 decimal places
for i in range(n):
    print('%.2f' % a[i], end=' ')

# Create an array 'b' with 5 logarithmically spaced values from 10^1 to 10^4 (default base=10)
b = np.logspace(1, 4, 5)

# Print the content of array 'b'
print('\n Content of array b is  \n', b)

# Loop through the first 4 elements of 'b' (note: 'b' has 5 elements, so the last one is skipped due to using len(a)=4)
for i in range(n):
    print('%.2f' % b[i], end=' ')

# Create an array 'c' with 4 logarithmically spaced values from 2^1 to 2^4 (explicit base=2)
c = np.logspace(1, 4, 4, base=2)

# Print the content of array 'c'
print('\n Content of array c is  \n', c)

# Loop through each element in 'c' and print it with 2 decimal places
for i in range(n):
    print('%.2f' % c[i], end=' ')
```
