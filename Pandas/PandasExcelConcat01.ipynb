```python
import pandas as pd

# This stops the output from wrapping to a new line
pd.set_option('display.expand_frame_repr', False)

# Optional: This ensures you see all columns without '...' in the middle
pd.set_option('display.max_columns', None)

# Your existing code
df1 = pd.read_excel('/content/EMP.xlsx', sheet_name='Sheet1')
df2 = pd.read_excel('/content/EMP.xlsx', sheet_name='Sheet2')

res = pd.concat([df1, df2], axis=1)
print('Concatenated Data:\n', res)
```
