```python
# Import necessary libraries: matplotlib for plotting and numpy (though not used here, possibly for future extensions)
import matplotlib.pyplot as plt
import numpy as np

# Define the data slices representing the proportion of employees in each department
slices =[50, 14, 24, 16]

# List of department names corresponding to the slices
# Note: 'Finanace' appears to be a typo and should likely be 'Finance'
dept =['Sales','Production','HR','Finanace']

# Custom colors for each pie slice
cols =['magenta','cyan','brown','gold']

# Create a pie chart:
# - slices: data values
# - labels: department names
# - startangle: rotate the start of the pie by 90 degrees
# - colors: custom color list
# - shadow: add a shadow effect for visual depth
# - explode: offset the second slice (Production) by 0.2 for emphasis
# - autopct: display percentage values with one decimal place
plt.pie(slices , labels=dept, startangle=90, colors=cols,shadow=True, explode=(0,0.2,0,0), autopct='%1.1f%%')

# Set the title of the chart
# Note: Title could be improved to 'Department-Wise Employee Distribution' for clarity
plt.title('Department Wise Employee')

# Display the plot
plt.show()
```
