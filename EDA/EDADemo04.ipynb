```python
# ================================================
# BASIC DATA LOADING AND EXPLORATION SCRIPT
# ================================================
# Author       : Your Name / Team
# Purpose      : Load, inspect, clean, and preview a car dataset
# Best Practice: Always keep the original data untouched using .copy()
# ================================================

import pandas as pd

# Improve readability in Jupyter/Colab notebooks by preventing line wrapping
pd.set_option('display.width', 1500)

# ------------------------------------------------------------------
# 1. LOAD THE DATA
# ------------------------------------------------------------------
# Reading CSV from Colab's default upload folder
# Always use full path or upload via files.upload() in production notebooks
df = pd.read_csv('/content/cars.csv')

# ------------------------------------------------------------------
# 2. CREATE A SAFE COPY
# ------------------------------------------------------------------
# Deep copy ensures that any changes to 'df' won't affect the original data
# This is a GOLDEN RULE in data cleaning — never modify the raw DataFrame directly!
df1 = df.copy()

# ------------------------------------------------------------------
# 3. BASIC INFORMATION ABOUT THE DATASET
# ------------------------------------------------------------------
# 3.1 Data types, non-null counts, memory usage
df.info()

# 3.2 Check shape (rows, columns) — no print() needed in notebooks (it auto-displays)
df.shape

# ------------------------------------------------------------------
# 4. DATA QUALITY CHECKS
# ------------------------------------------------------------------
# 4.1 Check for duplicate rows (entire row must match)
print("Total duplicate rows:", df.duplicated().sum())

# 4.2 Check for missing values (NaN) in each column
print('\n=== Sum of NA (missing values) in df ===')
print(df.isna().sum())

# ------------------------------------------------------------------
# 5. HANDLING MISSING VALUES
# ------------------------------------------------------------------
# Remove any row that has at least one missing value
# inplace=True modifies df directly (fast but irreversible)
df.dropna(inplace=True)

# Confirm cleaning worked
print('\n=== Sum of NA after dropna() ===')
print(df.isna().sum())

# ------------------------------------------------------------------
# 6. COMPARISON WITH ORIGINAL COPY (df1)
# ------------------------------------------------------------------
# The line below is INTENTIONALLY COMMENTED OUT so you can compare:
#    • df  → cleaned version
#    • df1 → still has original missing values
# This demonstrates the power of using .copy()

# df1.dropna(inplace=True)   # ← uncomment only if you want to clean df1 too

print('\n=== Sum of NA in df1 (original untouched copy) ===')
print(df1.isna().sum())   # df1 still has the original missing values

# ------------------------------------------------------------------
# 7. PREVIEW THE CLEANED DATA
# ------------------------------------------------------------------
# Always peek at the first few rows after cleaning
print("\n=== First 5 rows of cleaned DataFrame ===")
print(df.head())
```
