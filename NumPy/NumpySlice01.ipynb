```python
# Import NumPy library for efficient array operations
import numpy as np

# Create a 1D array 'arr' containing integers from 10 to 19 (arange excludes the stop value)
arr = np.arange(10, 20)
print('\n Content of Array  arr  \n', arr)  # Display the full array

# Slice the first 5 elements (indices 0 to 4) into a new array 'arr1'
# Syntax: arr[start:stop:step] → here step defaults to 1
arr1 = arr[0:5:]
print('\n Content of Array  arr  \n', arr1)  # Note: label says "arr" but actually printing arr1

# Get the number of elements in arr1
n = len(arr1)
print('\n Length of Array  arr1  \n', n)

# Initialize loop counter
i = 0

# Print each element of arr1 using a while loop (manual iteration)
while (i < n):
    print(arr1[i], end='  ')  # Print element followed by two spaces (no newline)
    i += 1  # Increment counter
```
