```python
# Import the NumPy library and alias it as 'np' for easier typing
import numpy as np

# Create two 1D NumPy arrays of the same shape (5 elements each)
a = np.array([10, 20, 30, 40, -45])
b = np.array([5, 10, 15, 20, 25])

# Element-wise Addition: Adds each element of 'a' to the corresponding element of 'b'
# Result: [10+5, 20+10, 30+15, 40+20, -45+25]
c = a + b

# Display the result of the addition
print('\n Sum of array a and b is \n' , c)

# Element-wise Subtraction: Subtracts each element of 'b' from the corresponding element of 'a'
# Result: [10-5, 20-10, 30-15, 40-20, -45-25]
c = a - b

# Display the result of the subtraction
print('\n Difference of array a and b is \n' , c)

```
