```python
import pandas as pd  # Import the pandas library for data manipulation and analysis

# Load the Excel file into a DataFrame
df = pd.read_excel('/content/EmployeesUpdated.xlsx')

# Set the display width option to 1000 for better readability of wide DataFrames
pd.set_option('display.width', 1000)

# Commented out: This option would prevent DataFrame representation from wrapping to multiple lines
# pd.set_option('display.expand_frame_repr', False)

# Equivalent SQL: SELECT * FROM EmployeesUpdated
# Print the entire DataFrame to display all employee details
print("\n Output of Complete Details \n ", df)

# Equivalent SQL: SELECT * FROM EmployeesUpdated WHERE Salary IN (42392, 60500, 129949, 130590)
# Note: Typo in the original comment: "slect" should be "select"
# Filter and print rows where Salary is in the specified list of values
print("\n Filtered Salary Details \n ", df[df['Salary'].isin([42392, 60500, 129949, 130590])])

# Equivalent SQL: SELECT * FROM EmployeesUpdated WHERE Salary NOT IN (42392, 60500, 129949, 130590)
# Note: Typo in the original comment: "slect" should be "select"
# Filter and print rows where Salary is not in the specified list of values (using the bitwise NOT operator ~)
print("\n Filtered Not Included  Salary Details \n ", df[~df['Salary'].isin([42392, 60500, 129949, 130590])])



```
