```python
import pandas as pd  # Import the pandas library for data manipulation and analysis

# Load the Excel file into a DataFrame
df = pd.read_excel('/content/EmployeesUpdated.xlsx')

# Set the display width option to 1000 for better readability of wide DataFrames
pd.set_option('display.width', 1000)

# Print the first 5 rows of the DataFrame for an overview
print(df.head())

# Filter and print all columns whose names start with 'S' using regex on axis=1 (columns)
# Note: Typo in the string: "name start with S" should be "names start with 'S'"
print('\n Filter out all columns whose name start with S \n', df.filter(regex='^S', axis=1))

# Filter and print all columns whose names end with 'e' using regex on axis=1 (columns)
# Note: Typo in the string: "name end with e" should be "names end with 'e'"
print('\n Filter out all columns whose name end with e \n ', df.filter(regex='e$', axis=1))

# Filter and print rows where the 'First Name' starts with 'Maria' (using regex '^Maria' in str.contains)
# Equivalent SQL: SELECT * FROM EmployeesUpdated WHERE First_Name LIKE 'Maria%'
print(df[df['First Name'].str.contains('^Maria')])
```
