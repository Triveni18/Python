```python
# Import necessary libraries for data manipulation and visualization
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set pandas display option to increase the width for better readability when printing DataFrames
pd.set_option('display.width', 1500)

# Load the 'mpg' dataset from seaborn, which contains fuel economy data for various cars
mpg = sns.load_dataset('mpg')

# Print a header and the entire mpg dataset to inspect the data
print("\n Content of MPG Dataset  \n" , mpg)

# Print a header for the horsepower vs acceleration graph
print("\n Graph horsepower VS acceleration \n")
# Print a subset of the data showing only 'horsepower' and 'acceleration' columns for reference
print(mpg[['horsepower','acceleration']])
# Create a scatterplot to visualize the relationship between horsepower and acceleration, using brown color for points
sns.scatterplot(x="horsepower", y="acceleration", data=mpg , color='brown')
# Display the plot
plt.show()
```
