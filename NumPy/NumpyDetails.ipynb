```python
import numpy as np

# This will print everything, including internal private attributes
all_objects = dir(np)
print(all_objects)
print(f"\nTotal objects found: {len(all_objects)}")
```

```python
import numpy as np

# This shows the intended public API
public_api = np.__all__
print(public_api)
print(f"\nTotal public objects: {len(public_api)}")
```

```python
import numpy as np

# Filter out anything starting with an underscore
clean_list = [obj for obj in dir(np) if not obj.startswith('_')]

for item in clean_list[:20]:  # Just printing the first 20 to save space
    print(item)

print(f"...\nTotal 'clean' objects: {len(clean_list)}")
```

```python
import numpy as np
import inspect

# Get all members and their types
members = inspect.getmembers(np)

functions = [name for name, obj in members if inspect.isfunction(obj)]
classes = [name for name, obj in members if inspect.isclass(obj)]


print(f"Found {len(functions)} functions.")
print(f"Found {len(classes)} classes.")

print(functions)

print(classes)
```
