```python
# Import necessary libraries for data visualization
import matplotlib.pyplot as plt
import numpy as np

# Generate sample data for the line plot
x = np.linspace(0, 10, 100)  # 100 points from 0 to 10
y = np.sin(x)  # Sine wave values

# Create a simple line plot
plt.figure(figsize=(8, 5))  # Set figure size
plt.plot(x, y, label='Sine Wave', color='red', linestyle='--', linewidth=2)  # Plot the line with custom style
plt.title('Simple Line Plot Demo')  # Add title
plt.xlabel('X-axis')  # Label x-axis
plt.ylabel('Y-axis')  # Label y-axis
plt.legend()  # Show legend
plt.grid(True)  # Add grid
plt.show()  # Display the plot

x = np.linspace(0, 10, 100)  # 100 points from 0 to 10
y = np.tan(x)  # Tan wave values

# Create a simple line plot
plt.figure(figsize=(8, 5))  # Set figure size
plt.plot(x, y, label='Tan Wave', color='brown', linestyle='--', linewidth=2)  # Plot the line with custom style
plt.title('Simple Line Plot Demo')  # Add title
plt.xlabel('X-axis')  # Label x-axis
plt.ylabel('Y-axis')  # Label y-axis
plt.legend()  # Show legend
plt.grid(True)  # Add grid
plt.show()  # Display the plot
```
