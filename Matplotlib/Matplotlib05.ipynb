```python
# Import pandas for data manipulation and matplotlib for plotting.
import pandas as pd
import matplotlib.pyplot as plt

# Set pandas display option to increase the width for better DataFrame printing.
pd.set_option('display.width', 2000)
pd.set_option('display.max_columns', None)
# Read the Excel file 'EmpDeptDetails.xlsx' from the specified path, selecting 'Sheet1'.
df = pd.read_excel('/content/EmpDeptDetails.xlsx', sheet_name='Sheet1')

# Filter the DataFrame to get data for the 'Sales' department.
sales_data = df[df['deptname'] == 'Sales']

# Print the Sales department data with a header.
print('\n Sales Department Data \n ', sales_data)

# Filter the DataFrame to get data for the 'Production' department.
prod_data = df[df['deptname'] == 'Production']

# Print the Production department data with a header.
print('\n Production Department Data \n ', prod_data)

# Create a new figure for the plot with a specified size (15 inches wide, 6 inches tall).
plt.figure(figsize=(15, 6))

# Create a bar chart for the Sales department salaries, using employee IDs on the x-axis,
# salaries on the y-axis, labeled 'Sales' with brown color and bar width of 0.5.
plt.bar(sales_data['empid'], sales_data['salary'], label='Sales', color='brown', width=0.5)

# Create a bar chart for the Production department salaries, using employee IDs on the x-axis,
# salaries on the y-axis, labeled 'Production' with orange color and bar width of 0.5.
plt.bar(prod_data['empid'], prod_data['salary'], label='Production', color='orange', width=0.5)

# Create bars for Sales data (re-assigning for labeling) and add labels to each bar with padding of 5.
bars = plt.bar(sales_data['empid'], sales_data['salary'], color='brown')
plt.bar_label(bars, padding=5)

# Create bars for Production data (re-assigning for labeling) and add labels to each bar with padding of 5.
bars = plt.bar(prod_data['empid'], prod_data['salary'], color='orange')
plt.bar_label(bars, padding=5)

# Add label to the x-axis: 'Employee ID'.
plt.xlabel('Employee ID')

# Add label to the y-axis: 'Salary'.
plt.ylabel('Salary')

# Set the title of the chart: 'Salary Comparison by Employee ID and Department'.
plt.title('Salary Comparison by Employee ID and Department')

# Display the legend to differentiate between departments.
plt.legend()

# Show the plotted chart.
plt.show()
```
