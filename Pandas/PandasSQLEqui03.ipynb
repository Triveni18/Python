```python
import pandas as pd  # Import the pandas library for data manipulation and analysis

# Load the Excel file into a DataFrame
df = pd.read_excel('/content/EmployeesUpdated.xlsx')

# Set the display width option to 1000 for better readability of wide DataFrames
pd.set_option('display.width', 1000)

# Commented out: This option would prevent DataFrame representation from wrapping to multiple lines
# pd.set_option('display.expand_frame_repr', False)

# Equivalent SQL: SELECT * FROM EmployeesUpdated
# Print the first 5 rows of the DataFrame (using head()), though the string suggests complete details
# Note: This prints only the head, not the entire DataFrame; if intending to print all, use print(df) instead
print("\n Output of Complete Details \n ", df.head())

# Equivalent SQL: SELECT * FROM EmployeesUpdated ORDER BY Salary DESC LIMIT 5
# Note: The comment above mentions LIMIT 5, matching the nlargest(5, 'Salary')
# Print the top 5 rows with the largest Salary values (descending order)
print("\n Top 5 Salary \n " ,df.nlargest(5, 'Salary'))

# Equivalent SQL approximation: SELECT * FROM (SELECT * FROM EmployeesUpdated ORDER BY Salary DESC LIMIT 10) ORDER BY Salary DESC LIMIT 5 OFFSET 5
# Retrieve the top 10 rows by Salary descending, then take the last 5 of those (effectively ranks 6-10)
print("\n Top next 5 Salary out of 10 Top Salary \n " ,df.nlargest(10, 'Salary').tail(5))
```
