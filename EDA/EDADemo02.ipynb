```python
# ================================================
# Handling Duplicate Rows in the Cars Dataset
# ================================================

import pandas as pd

# Set pandas display width for better readability of wide DataFrames
pd.set_option('display.width', 1500)

# Load the dataset into a DataFrame
# (Using 'df' as the variable name – good practice to keep it short and clear)
df = pd.read_csv('/content/cars.csv')

# Quick first look at the data
print("=== First 5 rows of the original dataset ===")
print(df.head())

# ------------------------------------------------
# 1. Check for Duplicate Rows
# ------------------------------------------------
# df.duplicated() returns a boolean Series (True if the row is a duplicate)
# .sum() counts how many True values → total number of duplicate rows
duplicate_count = df.duplicated().sum()
print(f"\nTotal duplicate rows: {duplicate_count}")

# ------------------------------------------------
# 2. View the Actual Duplicate Rows
# ------------------------------------------------
# df[df.duplicated()] filters and shows only the rows that are duplicates
# (By default, it excludes the first occurrence of each duplicate)
duplicates = df[df.duplicated()]

print("\n=== Sample of duplicate rows (first occurrence excluded) ===")
display(duplicates.head())          # display() works nicely in Jupyter/Colab notebooks

# ------------------------------------------------
# 3. Remove Duplicate Rows
# ------------------------------------------------
# drop_duplicates(keep='first') keeps the first occurrence and removes the rest
# We assign the cleaned data to a new DataFrame 'df1' (original 'df' remains unchanged)
df1 = df.drop_duplicates(keep='first')

print("\n=== First 5 rows after removing duplicates ===")
print(df1.head())

# ------------------------------------------------
# 4. Verify Duplicates Are Gone
# ------------------------------------------------
duplicate_count1 = df1.duplicated().sum()
print(f"\nTotal duplicate rows after cleaning: {duplicate_count1} ✅")

# Optional: If you want to update the original DataFrame instead:
# df.drop_duplicates(keep='first', inplace=True)
```
