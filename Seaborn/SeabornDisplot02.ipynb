```python
import pandas as pd     # Used for data manipulation (DataFrames)
import numpy as np      # Used for numerical operations and handling arrays
import seaborn as sns   # Used for creating beautiful statistical visualizations
import matplotlib.pyplot as plt # Needed to actually display and format the plots

# Increase the display width in the output console so tables don't wrap
pd.set_option('display.width', 1500)

# Load the dataset from a CSV file stored in the Google Colab 'content' folder
df = pd.read_csv('/content/auto-mpg.csv')

# Print the first 5 rows of the data to verify it loaded correctly
print(df.head(5))

# Create a distribution plot for the 'cylinders' column
# data=df: Tells seaborn to use your DataFrame
# x='cylinders': Specifies which column to plot on the horizontal axis
# bins=5: Groups the cylinder values into 5 bars
# kde=True: Adds a Kernel Density Estimate line (the smooth curve over the bars)
# color='brown': Sets the visual theme of the plot
sns.displot(data=df, x='cylinders', bins=5, kde=True, color='brown')

# Explicitly command the plot to show on the screen
plt.show()

```
