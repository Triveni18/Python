```python
# Import necessary libraries for data visualization and manipulation

import matplotlib.pyplot as plt  # For creating static, animated, and interactive visualizations
import numpy as np  # For numerical computations and array operations (though not directly used here)
import seaborn as sns  # For statistical data visualization built on matplotlib
import pandas as pd  # For data manipulation and analysis, especially with DataFrames

# Set Pandas display option to increase the console width for better output readability without wrapping
pd.set_option('display.width', 1500)

# Load the 'mpg' dataset from Seaborn, which contains fuel economy data for various cars
mpg = sns.load_dataset('mpg')

# Display the first few rows of the dataset to get an overview (in notebooks, mpg.head() alone would suffice for HTML output)
print(mpg.head())

# Count the occurrences of each unique value in the 'cylinders' column (e.g., how many cars have 4, 6, 8 cylinders)
cylinder_counts = mpg['cylinders'].value_counts()

# Count the occurrences of each unique value in the 'horsepower' column (e.g., how many cars have specific horsepower levels)
horse_counts = mpg['horsepower'].value_counts()

# Print the counts for cylinder types to show distribution
print("Vehicle counts per cylinder type:")
print(cylinder_counts)

# Print the counts for horsepower values to show distribution
print("Vehicle counts per horse power:")
print(horse_counts)

# Create a distribution plot (histogram with optional KDE) for the 'cylinders' column
# - data: The DataFrame to plot from
# - x: The column to plot on the x-axis
# - bins: Number of bins for the histogram (here, 5)
# - color: Color of the bars (red)
# - kde: If True, overlays a kernel density estimate curve
sns.displot(data=mpg, x='cylinders', bins=5, color='red', kde=True)

# Display the plot (necessary in scripts; in notebooks, it often shows automatically)
plt.show()

# Set the default Seaborn theme for styling future plots (note: this should ideally be called before plotting to affect the above displot)
sns.set()
```
