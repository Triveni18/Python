```python
import pandas as pd

# 1. Load the Excel file into a DataFrame
# Note: Ensure the file 'CityTemp.xlsx' is actually in the /content/ folder if using Colab or adjust the path for your local environment
# This assumes the file contains data with columns like 'City' and temperature-related columns (e.g., 'Temperature')
# Specifying sheet_name='Sheet1' loads data from that specific sheet; if omitted, it would load the first sheet by default
df = pd.read_excel('/content/CityTemp.xlsx', sheet_name='Sheet1')

# 2. Apply multiple aggregations to the entire DataFrame using agg()
# Note: Use lowercase 'max' instead of 'Max'—aggregation functions like 'max', 'min', 'count' are case-sensitive and must be lowercase
# This computes the maximum, minimum, and count for each column in the DataFrame
# The result is a DataFrame with these statistics as rows and original columns as columns
# For non-numeric columns, 'max' and 'min' return lexicographical max/min; 'count' returns non-null value counts
# If you want to apply this to specific columns, use df[['Column1', 'Column2']].agg(['max', 'min', 'count'])
# To apply different aggregations per column, use a dictionary like agg({'Column1': 'max', 'Column2': 'min'})
print(df.agg(['max', 'min', 'count']))
```
