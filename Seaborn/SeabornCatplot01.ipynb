```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

iris = sns.load_dataset('iris')
print('\n The Content of iris Dataset \n' ,iris.head())

print('\n  Cat Plot Graph based on iris data set species vs sepal_length \n')

sns.catplot(x='species',y='sepal_length',data=iris, color='brown')
plt.show()

print('\n  Cat Plot Graph based on iris data set species vs sepal_width \n')
sns.catplot(x='species',y='sepal_width',data=iris , color='red')
plt.show()

print('\n  Cat Plot Graph based on iris data set species vs petal_length \n')
sns.catplot(x='species',y='petal_length',data=iris , color='brown')
plt.show()
```
