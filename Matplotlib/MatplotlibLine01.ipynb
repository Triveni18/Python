```python
import matplotlib.pyplot as plt

years = [2015, 2016, 2017, 2018, 2019, 2020, 2021]
profits = [9, 10, 10.5, 8.9, 12.75, 7.51, 6.45]

# 1. Add the 'label' parameter here
plt.plot(years, profits, color='blue', label='Annual Profit')

plt.title('Profit Analysis')
plt.xlabel('Years')
plt.ylabel('Profits')

# 2. Call the legend function
plt.legend()

plt.grid(True)
plt.show()
```
