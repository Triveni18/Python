```python
# Import necessary libraries: pandas for data manipulation, matplotlib for plotting, and seaborn for enhanced visualizations
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set pandas display options to show all columns and rows without truncation, and set a specific width for better readability
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
pd.set_option('display.width', 500)

# Load the 'mpg' dataset from seaborn, which contains fuel economy data for various cars
mpg = sns.load_dataset('mpg')

# Print a header and display the first few rows of the MPG dataset for an overview
print('\n The content of MPG Data Set \n', mpg.head())

# Print a header for the first scatter plot with regression line
print('\n Scatter Plot With Regression Line \n ')

# Create a scatter plot with a linear regression line showing the relationship between horsepower and acceleration
sns.lmplot(x='horsepower', y='acceleration', data=mpg)

# Display the plot
plt.show()

# Print a header for the second scatter plot with regression line and color differentiation
print('\n Scatter Plot With Regression Line in Different Color \n ')

# Create a scatter plot with linear regression lines, colored by the 'origin' of the cars (e.g., USA, Japan, Europe)
sns.lmplot(x='horsepower', y='acceleration', data=mpg, hue='origin')

# Display the plot
plt.show()
```
