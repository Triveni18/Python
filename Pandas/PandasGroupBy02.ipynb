```python
import pandas as pd

# Load the Excel file into a DataFrame
# Note: Ensure the file path is correct for your environment (e.g., Google Colab or local setup)
# This assumes 'EMP.xlsx' contains employee data with columns like 'DEPARTMENT' and possibly others (e.g., employee names, salaries)
# Specifying sheet_name='Sheet2' loads data from that specific sheet; if omitted, it would load the first sheet by default
df = pd.read_excel('/content/EMP.xlsx', sheet_name='Sheet2')

# Group the DataFrame by the 'DEPARTMENT' column
# This creates a GroupBy object where rows are grouped by unique department values
# Note: GroupBy objects are iterable and can be used for aggregation, transformation, or iteration over groups
df1 = df.groupby('DEPARTMENT')
print(df1)  # Prints the GroupBy object (shows grouping info but not the actual data)
# print(df1)  # This line is already commented out and redundant since the print above is active

# Iterate over each group in the GroupBy object
# For each unique 'DEPARTMENT' value, 'rows' will be a DataFrame containing all rows in that department
# This is useful for inspecting or processing each group separately
# Note: The variable name 'DEPARTMENT' in the loop matches the grouping column for clarity
for DEPARTMENT, rows in df1:
    print(f"Department: {DEPARTMENT}")  # Print the department name for the current group
    print(rows)                        # Print the DataFrame subset for this group
    print("\n")                        # Add a newline for separation between groups

# Perform a groupby on 'DEPARTMENT' and apply the count() aggregation
# This counts the number of non-null values in each column for each unique 'DEPARTMENT'
# The result is a DataFrame with 'DEPARTMENT' as the index and counts for other columns
# Note: If you want 'DEPARTMENT' as a regular column instead of the index, use groupby(as_index=False)
# This is useful for getting frequency counts per department or seeing how many entries per department
df2 = df.groupby('DEPARTMENT').count()
print(df2)
```
