```python
import pandas as pd  # Import the pandas library for data manipulation and analysis

# Load the Excel file into a DataFrame
df = pd.read_excel('/content/EmployeesUpdated.xlsx')

# Set the display width option to 1000 for better readability of wide DataFrames
pd.set_option('display.width', 1000)

# Commented out: This option would prevent DataFrame representation from wrapping to multiple lines
# pd.set_option('display.expand_frame_repr', False)

# Equivalent SQL: SELECT * FROM EmployeesUpdated
# Print the entire DataFrame to display all employee details

print("\n Output of Complete Details \n ", df)

# Print the first 5 rows of the DataFrame (default behavior of head())
# Note: Typo in the string: "Ouput" should be "Output"
print("\n Ouput of Head \n " ,df.head())

# Equivalent SQL: SELECT * FROM EmployeesUpdated LIMIT 6
# Print the first 6 rows of the DataFrame
# Note: Typo in the string: "Ouput" should be "Output"
print("Ouput of Head  6 \n " ,df.head(6))

# Equivalent SQL: SELECT * FROM EmployeesUpdated WHERE Salary > 97308
# Filter and print rows where Salary is greater than 97308
print("Output Where salary is above 97308 \n ",df[df['Salary'] > 97308])

# Equivalent SQL: SELECT First_Name, Gender, Salary FROM EmployeesUpdated WHERE Salary = 97308
# Note: Typos in the original comment: "slect" should be "select", "Salry" should be "Salary"
# Filter rows where Salary equals 97308 and select specific columns: First Name, Gender, Salary
# Note: Typo in the string: "maultiple" should be "multiple"
print("\n Output with condition with maultiple columns name \n ",df[df['Salary']==97308] [['First Name','Gender','Salary']])

# Equivalent SQL: SELECT First_Name, Gender, Salary FROM EmployeesUpdated WHERE Salary = 130590 AND Team = 'Finance'
# Filter rows where Salary is 130590 and Team is 'Finance', then select specific columns
print("\n Output with condition with multiple columns name \n ",
      df[(df['Salary'] == 130590) & (df['Team'] == 'Finance')][['First Name', 'Gender', 'Salary']])

# Equivalent SQL: SELECT Gender, Salary, COUNT(*) FROM EmployeesUpdated GROUP BY Gender, Salary
# Note: Typo in the original comment: "salay" should be "salary"
# Group by Gender and Salary, count the occurrences, and reset index to create a DataFrame with a 'Count' column
print(df.groupby(['Gender', 'Salary']).size().reset_index(name='Count'))

# Filter for male employees and sort by Salary in descending order
# Note: This does not correspond to the previous SQL comment; it's an additional operation
print(df[df['Gender']=='Male'].sort_values(by='Salary',ascending=False))


```
