```python
# Import the numpy library and alias it as 'np' for convenience
import numpy as np

# Create a numpy array 'arr' containing the integers 10, 20, 30, 40, and 50
arr = np.array([10, 20, 30, 40, 50])

# Print a newline, followed by the string "Content of Array", another newline, and then the contents of the array
print('\n Content of Array  \n', arr)
```
