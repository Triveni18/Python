```python
import numpy as np

# Create a 1D (one-dimensional) array with 6 elements
arr1 = np.array([10, 20, 30, 40, 50, 60])
print('Content of array arr1', arr1)

# Change the shape of the array to 2 rows and 3 columns
# Note: The total number of elements must remain the same (2 * 3 = 6)
arr2 = arr1.reshape(2, 3)
print('Content of array arr2', arr2)

# Reshape back into a 2D array with 1 row and 6 columns
# Note: This is different from a 1D array because it has two sets of brackets [[]]
arr3 = arr2.reshape(1, 6)
print('Content of array arr3', arr3)

# Collapse the multi-dimensional array back into a simple 1D array
# This "unrolls" the matrix into a single flat line of data
arr4 = arr2.flatten()
print('Content of array arr4', arr4)
```
