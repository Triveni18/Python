```python
import pandas as pd

# Load the Excel file into a DataFrame
# Note: Ensure the file path is correct for your environment (e.g., Google Colab or local setup)
# This assumes 'EMP.xlsx' contains employee data with columns like 'SALARY' and possibly others
df = pd.read_excel('/content/EMP.xlsx')

# 1. Using mask: This method replaces values in the DataFrame where the condition is True
# Here, for rows where 'SALARY' <= 49000, the entire row is replaced with 'Lowest'
# Warning: This applies to the whole DataFrame, not just the 'SALARY' column. If you intend to replace only in 'SALARY',
# use df['SALARY'] = df['SALARY'].mask(df['SALARY'] <= 49000, 'Lowest') instead
df1 = df.mask(df['SALARY'] <= 49000, 'Lowest')
print(df1)

# Create a copy of the original DataFrame to modify only the 'SALARY' column using mask
# This replaces values in 'SALARY' <= 49000 with 'Lowest', keeping higher salaries unchanged
# Note: This changes the 'SALARY' column dtype to object since it mixes numbers and strings
df2 = df.copy()
df2['SALARY'] = df['SALARY'].mask(df['SALARY'] <= 49000, 'Lowest')
print(df2)


# OR

# 2. Using where: This method keeps original values where the condition is True, and replaces them where False
# Here, for rows where 'SALARY' <= 49000, keep the original row; otherwise (SALARY > 49000), replace the entire row with 'Lowest'
# Note: The comment above this line in the original code was incorrect—it described the opposite condition
# Similar warning: This affects the entire DataFrame. To target only 'SALARY', use df['SALARY'] = df['SALARY'].where(df['SALARY'] <= 49000, 'Lowest')
df1 = df.where(df['SALARY'] <= 49000, 'Lowest')
print(df1)

# Create a copy of the original DataFrame to modify only the 'SALARY' column using where
# This keeps original values in 'SALARY' where <= 49000, and replaces higher salaries with 'Lowest'
# Note: This effectively labels higher salaries as 'Lowest', which may not align with the label's intent (consider adjusting the condition if aiming to label low salaries)
# Also, this changes the 'SALARY' column dtype to object since it mixes numbers and strings
df3 = df.copy()
df3['SALARY'] = df['SALARY'].where(df['SALARY'] <= 49000, 'Lowest')
print(df3)
#
```
