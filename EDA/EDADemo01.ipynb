```python
# ================================================
# Basic Exploratory Data Analysis (EDA) on Cars Dataset
# ================================================

# Import required libraries
import pandas as pd

# Set pandas display width for better readability of wide DataFrames in output
pd.set_option('display.width', 1500)

# Load the cars dataset from CSV file (Google Colab path)
# This creates a pandas DataFrame called 'cars'
cars = pd.read_csv("/content/cars.csv")

# ------------------------------------------------
# 1. Shape Attribute
# ------------------------------------------------
# Returns a tuple (number_of_rows, number_of_columns)
print('\n Result of Shape Attribute \n')
print(cars.shape)

# ------------------------------------------------
# 2. Info Method
# ------------------------------------------------
# Gives a concise summary of the DataFrame:
# - Column names, non-null counts, data types, and memory usage
print('\n Result of Info Method \n')
print(cars.info())

# ------------------------------------------------
# 3. Columns Attribute
# ------------------------------------------------
# Returns the column labels of the DataFrame as an Index object
print('\n Result of Columns Attribute  \n')
print(cars.columns)

# ------------------------------------------------
# 4. Describe Method
# ------------------------------------------------
# Generates descriptive statistics for numeric columns:
# count, mean, std, min, 25%, 50%, 75%, max
print('\n Result of Describe Method  \n')
print(cars.describe())

# ------------------------------------------------
# 5. Head Method
# ------------------------------------------------
# Returns the first 5 rows by default (useful to see the top of the data)
print('\n Result of head Method  \n')
print(cars.head())

# ------------------------------------------------
# 6. Tail Method
# ------------------------------------------------
# Returns the last 5 rows by default (useful to see the bottom of the data)
print('\n Result of Tail Method  \n')
print(cars.tail())
```
