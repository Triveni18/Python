```python
# Import the numpy library for numerical operations
import numpy as np

# Create an array 'a' with values from 0 to 9 (default start=0, step=1, stop=10 excludes 10)
a = np.arange(10)

# Print the content of array 'a'
print('\n The content of array a ', a)

# Create an array 'b' with values from 10 to 19 (start=10, stop=20 excludes 20, default step=1)
b = np.arange(10, 20)

# Print the content of array 'b'
print('\n The content of array b ', b)

# Create an array 'c' with values from 10 to 18 in steps of 2 (start=10, stop=20 excludes 20, step=2)
c = np.arange(10, 20, 2)

# Print the content of array 'c'
print('\n The content of array c ', c)

# Create an array 'd' with values from 10 down to 2 in steps of -1 (start=10, stop=1 excludes 1, step=-1)
d = np.arange(10, 1, -1)

# Print the content of array 'd'
print('\n The content of array d ', d)

# Create an array 'e' with values from 10 down to 1 in steps of -1 (start=10, stop=0 excludes 0, but since step=-1, it includes 1)
e = np.arange(10, 0, -1)

# Print the content of array 'e'
print('\n The content of array e ', e)
```
