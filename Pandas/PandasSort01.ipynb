```python
import pandas as pd

# Load the Excel file into a DataFrame
# If no sheet_name is specified, it defaults to the first sheet
df = pd.read_excel('/content/EMP.xlsx')

# Display the original, unsorted data
print("Original Data:")
print(df)

# Sort the DataFrame by 'Date of Joining' (DOJ) in ascending order (default)
# The result is assigned back to 'df' to save the changes
df = df.sort_values('DOJ')

print("\nData sorted by DOJ (Oldest to Newest):")
print(df)

# Sort by multiple columns: first by 'DOJ', then by 'SALARY'
# ascending=False applies to both columns (Newest DOJ and Highest Salary first)
df = df.sort_values(['DOJ', 'SALARY'], ascending=False)

print("\nData sorted by DOJ and Salary (Descending):")
print(df)
```
