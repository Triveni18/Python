```python
# Import the pandas library, which is used for data manipulation and analysis.
# Pandas provides data structures like DataFrames for handling tabular data efficiently.
import pandas as pd

# Load the Excel file from the specified path and select the specific sheet.
# Ensure the file path is correct for your Google Colab or local environment, and that 'EMP.xlsx' is uploaded if necessary.
# This reads the Excel file into a DataFrame object named 'df', which represents the data in a tabular format.
# The 'sheet_name' parameter specifies which sheet in the Excel workbook to load (in this case, "Sheet1").
df = pd.read_excel('/content/EMP.xlsx', sheet_name="Sheet1")

# Filter the DataFrame to include only rows where the 'SALARY' column is greater than 65000, and select specific columns: 'EMPID', 'EMPNAME', 'DOJ'.
# This uses boolean indexing: df['SALARY'] > 65000 creates a boolean mask, and df[...] applies it to select matching rows.
# The [['EMPID','EMPNAME','DOJ']] specifies the columns to include in the resulting DataFrame.
dec = df[['EMPID','EMPNAME','DOJ']][df['SALARY'] > 65000]

# Display the filtered results by printing the DataFrame subset.
print(dec)

# Filter the DataFrame to include only rows where the 'SALARY' column is less than or equal to 50000, and select specific columns: 'EMPID', 'EMPNAME', 'DOJ'.
dec = df[['EMPID','EMPNAME','DOJ']][df['SALARY'] <= 50000]

# Print a descriptive message followed by the filtered DataFrame.
# The '\n' adds a newline for better readability in the output.
print('Details of People whose salary is less than 50000 \n ', dec)

# Filter the DataFrame to include only rows where the 'SALARY' is equal to the maximum salary in the dataset, and select specific columns: 'EMPID', 'EMPNAME', 'DOJ'.
# df['SALARY'].max() computes the highest value in the 'SALARY' column.
dec = df[['EMPID','EMPNAME','DOJ']][df['SALARY'] == df['SALARY'].max()]

# Print a descriptive message followed by the filtered DataFrame for the maximum salary.
print('Details of People whose salary is is equal to max  \n ', dec)

# Filter the DataFrame to include only rows where the 'SALARY' is equal to the minimum salary in the dataset, and select specific columns: 'EMPID', 'EMPNAME', 'DOJ'.
# df['SALARY'].min() computes the lowest value in the 'SALARY' column.
dec = df[['EMPID','EMPNAME','DOJ']][df['SALARY'] == df['SALARY'].min()]

# Print a descriptive message followed by the filtered DataFrame for the minimum salary.
print('Details of People whose salary is is equal to min  \n ', dec)

# Filter the DataFrame to include only rows where the 'SALARY' is less than or equal to the average salary, and select specific columns: 'EMPID', 'EMPNAME', 'DOJ'.
# df['SALARY'].mean() calculates the mean (average) value of the 'SALARY' column.
dec = df[['EMPID','EMPNAME','DOJ']][df['SALARY'] <= df['SALARY'].mean()]

# Print a descriptive message followed by the filtered DataFrame for salaries below or equal to the average.
print('Details of People whose salary is less than  Avg  \n ', dec)

# Filter the DataFrame to include only rows where the 'SALARY' is greater than or equal to the average salary, and select specific columns: 'EMPID', 'EMPNAME', 'DOJ'.
dec = df[['EMPID','EMPNAME','DOJ']][df['SALARY'] >= df['SALARY'].mean()]

# Print a descriptive message followed by the filtered DataFrame for salaries above or equal to the average.
print('Details of People whose salary is greater than  Avg  \n ', dec)
```
