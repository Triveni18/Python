```python
# Import necessary libraries: pandas for data manipulation, matplotlib for plotting, and seaborn for enhanced visualizations
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set pandas display options to show more columns and wider output for better readability in the console or notebook
pd.set_option('display.max_columns', 500)
pd.set_option('display.width', 1000)

# Load the built-in 'iris' dataset from seaborn, which contains measurements of iris flowers
iris = sns.load_dataset('iris')

# Print a header and the first few rows of the iris dataset to inspect its content
print('\n The Content of iris Data Set \n' , iris.head())

# Create a subset of the iris dataset: select the first 10 rows and the first 4 columns (which are the numerical features)
iris1 = iris.iloc[0:10 , 0:4]

# Print a header and the first few rows of the iris1 subset to verify the selection
print('\n The Content of iris1 Data Set \n' , iris1.head())

# Print a header for the first heatmap visualization
print('\n Heat Map \n')

# Generate a heatmap for the iris1 data with annotations (values displayed in cells) using default colormap
sns.heatmap(iris1 , annot=True)

# Display the plot (necessary in non-interactive environments, though often implicit in notebooks)
plt.show()

# Print a header for the second heatmap visualization
print('\n Heat Map with CMAP \n')

# Generate another heatmap for the iris1 data with annotations, but using the 'coolwarm' colormap for better contrast
sns.heatmap(iris1 , cmap='coolwarm',annot=True)

# Display the second plot
plt.show()
```
