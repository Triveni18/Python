```python
# Import necessary libraries:
# - pandas for data manipulation (though not directly used here beyond potential future needs)
# - matplotlib.pyplot for displaying plots
# - seaborn for data visualization and loading datasets
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the Iris dataset from seaborn's built-in datasets
# This dataset contains measurements of iris flowers from three species: setosa, versicolor, and virginica
iris = sns.load_dataset('iris')

# Print the first few rows of the dataset to inspect its content
print('\n The content of iris dataset   \n' , iris.head())

# Print a header for the first categorical plot
print('\n Cat plot based on iris dataset species vs sepal_length type violin  \n')

# Create a violin plot using seaborn's catplot
# - x='species': Categories on the x-axis (iris species)
# - y='sepal_length': Values on the y-axis (sepal length measurements)
# - data=iris: The dataset to use
# - kind='violin': Type of plot (violin plot shows distribution density)
# - color='brown': Sets the color of the violins
sns.catplot(x='species', y='sepal_length', data=iris, kind='violin', color='brown')

# Display the plot using matplotlib
plt.show()

# Print a header for the second categorical plot
print('\n Cat plot based on iris dataset species vs petal_length type violin  \n')

# Create another violin plot, this time comparing species to petal length
# Similar parameters as above, but y='petal_length' for petal length measurements
sns.catplot(x='species', y='petal_length', data=iris, kind='violin', color='brown')

# Display the plot
plt.show()

# Print a header for the third categorical plot
print('\n Cat plot based on iris dataset species vs sepal_length type boxen  \n')

# Create a boxen plot (enhanced box plot showing more quantiles)
# - kind='boxen': Type of plot for detailed distribution summary
# Other parameters same as the first plot
sns.catplot(x='species', y='sepal_length', data=iris, kind='boxen', color='brown')

# Display the plot
plt.show()
```
