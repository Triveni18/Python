```python
# Import necessary libraries: pandas for data manipulation, matplotlib for plotting, and seaborn for enhanced visualizations
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set pandas display option to increase the width for better readability when printing DataFrames
pd.set_option('display.width', 1500)

# Load the 'mpg' dataset from seaborn, which contains fuel economy data for various cars
mpg = sns.load_dataset('mpg')

# Print the entire content of the MPG dataset with a header for clarity
print('\n Content of MPG Data Set  \n', mpg)

# Print a header for the selected columns: horsepower and acceleration
print('\n The Horsepower and Acceleration \n')

# Extract and print only the 'horsepower' and 'acceleration' columns from the dataset
print(mpg[['horsepower', 'acceleration']])

# Print a header for the first line plot
print('\n Line Plot Graph based on Horsepower and Acceleration \n ')

# Create a simple line plot showing the relationship between horsepower and acceleration
sns.lineplot(x='horsepower', y='acceleration', data=mpg)

# Display the plot
plt.show()

# Print a header for the second line plot with color differentiation
print('\n Line Plot Graph based on Horsepower and Acceleration in different color \n ')

# Create a line plot showing the relationship between horsepower and acceleration, colored by the 'origin' of the cars (e.g., USA, Japan, Europe)
sns.lineplot(x='horsepower', y='acceleration', data=mpg, hue='origin')

# Display the plot
plt.show()
```
