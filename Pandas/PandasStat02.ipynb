```python
# Import the pandas library, which is used for data manipulation and analysis.
# Pandas provides data structures like DataFrames for handling tabular data efficiently.
import pandas as pd

# Load the Excel file from the specified path and select the specific sheet.
# Ensure the file path is correct for your Google Colab or local environment.
# This reads the Excel file into a DataFrame object named 'df', which represents the data in a tabular format.
# The 'sheet_name' parameter specifies which sheet in the Excel workbook to load (in this case, "Sheet1").
df = pd.read_excel('/content/EMP.xlsx', sheet_name="Sheet1")

# Generate descriptive statistics for the DataFrame.
# The describe() method computes summary statistics like count, mean, std deviation, min, max, and quartiles for numerical columns.
# It ignores non-numeric columns by default.
dec = df.describe()

# Print the descriptive statistics to the console.
# This outputs the summary in a readable tabular format.
print(dec)
```
