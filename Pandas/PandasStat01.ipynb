```python
import pandas as pd  # Import the pandas library, which is used for data manipulation and analysis

# Load the Excel file from the specified path and select the specific sheet
# Ensure the file path is correct for your Google Colab or local environment
# This reads the Excel file into a DataFrame object named 'df'
df = pd.read_excel('/content/EMP.xlsx', sheet_name="Sheet1")

# Calculate the maximum salary from the 'SALARY' column
ma = df['SALARY'].max()

# Calculate the minimum salary from the 'SALARY' column
mi = df['SALARY'].min()

# Calculate the average (mean) salary from the 'SALARY' column
avg = df['SALARY'].mean()

# Calculate the sum of all salaries from the 'SALARY' column
sum = df['SALARY'].sum()

# Count the number of non-null entries in the 'SALARY' column
co = df['SALARY'].count()

# Print the maximum salary
print("Maximum Salary:", ma)

# Print the minimum salary
print("Minimum Salary:", mi)

# Print the average salary
print("Average Salary:", avg)

# Print the sum of salaries
print("Sum of Salaries:", sum)

# Print the count of salaries
print("Count of Salaries:", co)
```
