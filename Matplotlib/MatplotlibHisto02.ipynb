```python
import pandas as pd  # Import pandas library for data manipulation and analysis
import matplotlib.pyplot as plt  # Import matplotlib's pyplot for creating visualizations

# Assuming your CSV is loaded correctly
df = pd.read_csv('/content/employee_age.csv', encoding='latin1')  # Load the CSV file into a pandas DataFrame; 'encoding' handles special characters if needed

bins = [20, 30, 40, 50]  # Define the bin edges for the histogram (groups ages into 20-30, 30-40, 40-50)

emp_age = df['age']  # Extract the 'age' column from the DataFrame as a Series for plotting

# Key change: align='mid' is the default, but rwidth affects the visual gap.
# If you want them centered on the tick marks, use the code below:
plt.hist(emp_age, bins=bins, histtype='bar', rwidth=0.4, color='brown', align='mid', edgecolor='black')
# Plot a histogram:
# - emp_age: data to bin
# - bins: bin edges
# - histtype='bar': standard bar histogram
# - rwidth=0.4: relative width of bars (creates gaps between bars)
# - color='brown': fill color of bars
# - align='mid': aligns bars to the middle of the bin
# - edgecolor='black': outline color for bars

plt.title('Employee Age Histogram')  # Set the title of the plot
plt.xlabel('Age')  # Set the x-axis label
plt.ylabel('Frequency')  # Set the y-axis label
plt.xticks(bins)  # This ensures the x-axis labels match your bin edges (places ticks at 20, 30, 40, 50)
#plt.grid(axis='y', alpha=0.75)  # Optional: Add a grid on the y-axis with transparency (commented out)
plt.show()  # Display the plot

```
