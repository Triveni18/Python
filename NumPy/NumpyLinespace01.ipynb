```python
# Import the NumPy library and alias it as np for convenience
import numpy as np

# Create an array 'arr' with 5 evenly spaced values starting from 0 to 10 (inclusive)
arr = np.linspace(0,10,5)

# Create an array 'arr1' with 5 evenly spaced values starting from 1 to 10 (inclusive)
arr1 =np.linspace(1,10, 5)

# Print the contents of the array 'arr'
print('\n The Value of Array arr \n' , arr)

# Print the contents of the array 'arr1'
print('\n The Value of Array arr 1 \n',arr1)
```
