```python
import pandas as pd
import numpy as np

pd.set_option('display.width', 1500)

df = pd.read_csv('/content/Automobile.csv')

# Handle missing values ('?') and convert to numeric
df = df.replace('?', np.nan)
numeric_cols = ['mpg', 'cylinders', 'displacement', 'horsepower', 'weight', 'acceleration', 'model_year', 'origin']
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Compute correlations with 'weight' using different methods
print("Pearson correlations (default) with 'weight':")
print(df[numeric_cols].corr(method='pearson')['weight'])

print("\nSpearman correlations with 'weight':")
print(df[numeric_cols].corr(method='spearman')['weight'])

print("\nKendall correlations with 'weight':")
print(df[numeric_cols].corr(method='kendall')['weight'])
```
