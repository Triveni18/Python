```python
import pandas as pd

# Load the Excel file
df = pd.read_excel('/content/EMPNA.xlsx')

print("Original Data with NaNs:")
print(df)

# OPTION 1: Replace ALL NaNs with 0 (Your current approach)
df1 = df.fillna(0)
print("\nAll NaNs replaced with 0:")
print(df1)

# OPTION 2: Targeted replacement (Better Practice)
# We can use a dictionary to specify different values for different columns
df2 = df.fillna({
    'SALARY': 'Salary Missing',           # Replace missing salary with 0
    'EMPNAME': 'Name not written',     # Replace missing names with 'Unknown'
    'DOJ': '00-00-00'   # Replace missing department with 'Unassigned'
})

print("\nTargeted Fill (Cleaned):")
print(df2)

df3 = df.dropna()
print("\nRows with NaNs removed:")
print(df3)

```
